```python
import nltk
from nltk.tokenize import word_tokenize
nltk.download("punk")
nltk.download("punkt_tab")
```

```python
text = "NLP is intresting! i love learning AI."
tokens = word_tokenize(text)
print("originalText:", text)
print("tokens:", tokens)
print("Total Tokens:", len(tokens))
```

```python
from numpy import vectorize
from sklearn.feature_extraction.text import CountVectorizer
documents = [
    " I Love machine learning", "I love python programming", "I love AI"]
vectorizer = CountVectorizer()
print("Vocabulary:")
X = vectorizer.fit_transform(documents)
print(vectorizer.get_feature_names_out())
print("\nBag of Words matrix")
print(X.toarray())

```

```python
from nltk import ngrams
from nltk.tokenize import wordpunct_tokenize
text = "I love machine learning"# input sentence
words = wordpunct_tokenize(text) # Tokenize the sentence
# Unigram (N = 1)
unigrams = list(ngrams(words, 1))
print("Unigrams:", unigrams)
# Bigram ( N=2)
bigrams = list(ngrams(words, 2))
print("Bigrams:", bigrams)
# Trigram (N=3)
trigrams = list(ngrams(words, 3))
print("Trigrams:", trigrams)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
documents = [
    "I love machine learning",
    "I love python programming",
    "I love AI"
]
# Create TF-IDF model
vectorizer = TfidfVectorizer()
# Generate the TF-IDF vectors for the documents
X = vectorizer.fit_transform(documents)
# Display Vocabulary
print("Vocabulary:")
print(vectorizer.get_feature_names_out())
# Display TF-IDF values
print("\nTF-IDF Matrix")
print(X.toarray().round(2))
```
